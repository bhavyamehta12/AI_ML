# 🟡 Intermediate Q13 — Set Intersection

Find common skills: `{'Python','HTML','CSS','JavaScript'}` and `{'Python','Java','CSS','SQL'}`.


```python
# Write your solution here
s1={'Python','HTML','CSS','JavaScript'}
s2={'Python','Java','CSS','SQL'}
l=[]
for i in s1:
    for j in s2:
        if i==j:
            l.append(i)
print(l)         
```
